```python
# Variables provisionales

promedio_salario = df["salary_usd"].mean()

max_productividad = df["productivity_score"].max()

print("Promedio salario:", promedio_salario)
print("Máxima productividad:", max_productividad)
```

```python
# Aplicaciones de numpy

# Promedio
print("Promedio de horas trabajadas:")
print(np.mean(df["weekly_hours_worked"]))


# Mediana
print("Mediana de salarios:")
print(np.median(df["salary_usd"]))


# Maximo
print("Productividad maxima:")
print(np.max(df["productivity_score"]))


# Minimo
print("Menores horas de sueño:")
print(np.min(df["sleep_hours_avg"]))


# Sumatoria
print("Total de tareas completadas:")
print(np.sum(df["tasks_completed"]))


# Desviacion Estandar
print("Desviacion estandar de satisfaccion laboral:")
print(np.std(df["job_satisfaction"]))

```

```python
#Filtros

# Empleados con alta productividad
alta_productividad = df[df["productivity_score"] >= 95]
print("Cantidad con productividad alta:", alta_productividad.shape[0])
print(alta_productividad.head())

# Empleados con muchas horas trabajadas
muchas_horas = df[df["weekly_hours_worked"] > 50]
print("Cantidad que trabaja más de 50 horas:", muchas_horas.shape[0])
```

```python
# Ordenamientos
print("\nTop 10 salarios más altos:")
print(df.sort_values("salary_usd", ascending=False).head(10))

print("\nTop 10 mayor productividad:")
print(df.sort_values("productivity_score", ascending=False).head(10))
```

```python
# Agrupaciones
print("\nPromedio de salario por industria:")
print(df.groupby("industry")["salary_usd"].mean().sort_values(ascending=False))

print("\nPromedio de productividad por modo de trabajo:")
print(df.groupby("work_mode")["productivity_score"].mean().sort_values(ascending=False))

print("\nCantidad de empleados por país:")
print(df.groupby("country")["employee_id"].count().sort_values(ascending=False))
```

```python
#Pivots

# Promedio de salario por industria y modo de trabajo
tabla_pivote_salario = df.pivot_table(
    values="salary_usd",
    index="industry",
    columns="work_mode",
    aggfunc="mean"
)

print(tabla_pivote_salario)

# Promedio de productividad por país y burnout
tabla_pivote_prod = df.pivot_table(
    values="productivity_score",
    index="country",
    columns="burnout_level",
    aggfunc="mean"
)

print(tabla_pivote_prod)
```

```python
#Combinacion de datasets
parte1 = df[["employee_id", "country", "industry", "salary_usd"]].sample(frac=0.8, random_state=42)
parte2 = df[["employee_id", "work_mode", "productivity_score", "burnout_level", "job_satisfaction"]]
```

```python
# Aplicacion de Inner Joinm
merge_inner = pd.merge(parte1, parte2, on="employee_id", how="inner")
print("INNER JOIN:", merge_inner.shape)
print(merge_inner.head())
```

```python
# Aplicacion de Outer Join
merge_outer = pd.merge(parte1, parte2, on="employee_id", how="outer")
print("OUTER JOIN:", merge_outer.shape)
print(merge_outer.head())
```

```python
# Gráfico de barras
industria_count = df["industry"].value_counts()

plt.figure(figsize=(8,5))
ax = industria_count.plot(kind="bar")

plt.title("Cantidad de empleados por industria")
plt.xlabel("Industria")
plt.ylabel("Cantidad de empleados")
plt.xticks(rotation=45)

bars = ax.patches
for bar in bars:
    altura = bar.get_height()

    ax.text(
        bar.get_x() + bar.get_width()/2,
        altura + 0.5,
        str(int(altura)),
        ha='center',
        va='bottom',
        fontsize=10
    )

plt.tight_layout()
plt.show()
```

```python
#Modelo de regresion
promedio = df.groupby("year")["job_satisfaction"].mean()


X = promedio.index.values.reshape(-1, 1)
y = promedio.values


modelo = LinearRegression()


modelo.fit(X, y)


prediccion = modelo.predict([[2026]])

print("Predicción para 2026:", prediccion)
```

```python
#Grafico de regresion
plt.figure(figsize=(12,6))


plt.scatter(X, y)
plt.plot(X, modelo.predict(X))

#For para mostrar valores
for i in range(len(X)):
    plt.text(
        X[i][0],
        y[i],
        f"{y[i]:.2f}",
        fontsize=10
    )

#Resultado de le prediccion
plt.scatter(2026, prediccion)

plt.text(
    2026,
    prediccion[0],
    f"{prediccion[0]:.2f}",
    fontsize=9
)

# Títulos
plt.title("Promedio de Satisfacción Laboral por Año")
plt.xlabel("Year")
plt.ylabel("Satisfaccion laboral")

plt.xticks(range(2018, 2027))

plt.grid()


plt.show()
```
